### You have built an AI

```python
import csv
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

with open("students.csv") as f:
    data = list(csv.reader(f))[1:]
X = [[float(v) for v in r[:3]] for r in data]
y = [int(r[3]) for r in data]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
model = LogisticRegression()
model.fit(X_train, y_train)

new_students = [[8, 2, 75],          # one row per student, same 3 features, same order
                [2, 8, 60]]
print("prediction:          ", model.predict(new_students))
print("probability of pass: ", model.predict_proba(new_students)[:, 1].round(3))
```

## 7. Evaluate: the 20 students it never saw

```python
import csv
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

with open("students.csv") as f:
    data = list(csv.reader(f))[1:]
X = [[float(v) for v in r[:3]] for r in data]
y = [int(r[3]) for r in data]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
model = LogisticRegression()
model.fit(X_train, y_train)

pred = model.predict(X_test)          # the model's answers for the 20 held-back students
print("predicted:", pred.tolist())
print("truth:    ", y_test)
print("test accuracy:", accuracy_score(y_test, pred))
print(confusion_matrix(y_test, pred))
```

## 8. The whole program, part by part

```python
import csv
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Data: read X and y, then split
with open("students.csv") as f:
    rows = list(csv.reader(f))[1:]
X = [[float(v) for v in r[:3]] for r in rows]
y = [int(r[3]) for r in rows]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

# 2. Model: choose the form of the answer
model = LogisticRegression()

# 3. Algorithm: train it, which fills in the weights
model.fit(X_train, y_train)

# 4. Evaluation: score it on the students it never saw
pred = model.predict(X_test)
print("test accuracy:", accuracy_score(y_test, pred))
```
