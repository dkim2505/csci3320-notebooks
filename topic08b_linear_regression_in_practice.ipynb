## The whole model, in three lines

```python
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

model = LinearRegression()          # 1. choose
model.fit(X_train, y_train)         # 2. train    <- Topic 8a's loop, done for you
pred = model.predict(X_test)        # 3. predict

print("first five predictions:", pred[:5].round(1))
print("first five actual     :", y_test[:5])
print(f"\nintercept: {model.intercept_:.2f}")
print(f"one coefficient per feature: {len(model.coef_)}")
```

## Scoring the model

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

model = LinearRegression().fit(X_train, y_train)
pred  = model.predict(X_test)

rmse = np.sqrt(mean_squared_error(y_test, pred))
r2   = r2_score(y_test, pred)

print(f"RMSE : {rmse:.2f}")
print(f"R2   : {r2:.4f}")

# the same two numbers, written out, so the formulas are not magic
print(f"\nby hand RMSE : {np.sqrt(np.mean((y_test - pred) ** 2)):.2f}")
ss_res = ((y_test - pred) ** 2).sum()
ss_tot = ((y_test - y_test.mean()) ** 2).sum()
print(f"by hand R2   : {1 - ss_res / ss_tot:.4f}")
```

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)

lazy = np.full_like(y_test, y_train.mean())          # always predict the average
fit  = LinearRegression().fit(X_train, y_train).predict(X_test)

for name, p in [("always the average", lazy), ("linear regression", fit)]:
    print(f"{name:20}  RMSE {np.sqrt(mean_squared_error(y_test, p)):6.2f}"
          f"   R2 {r2_score(y_test, p):7.4f}")
```

## What to remember

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
model = LinearRegression().fit(X_train, y_train)
pred  = model.predict(X_test)
print(np.sqrt(mean_squared_error(y_test, pred)), r2_score(y_test, pred))
```
